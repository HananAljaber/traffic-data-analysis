```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Traffic.csv")
```

```python
df.head()
```

```python
df.tail()
```

- اولا Time نحوله لتاريخ 

```python
df.info()
```

```python

df["Time"] = pd.to_datetime(df["Time"]).dt.strftime("%I:%M %p")
```

```python
df["Hour"] = pd.to_datetime(df["Time"]).dt.hour
```

### ما أكثر ساعة ازدحامًا؟

```python
df.groupby("Hour")["Total"].mean().index
df.groupby("Hour")["Total"].mean().values
```

### تحليل ساعات الذروة المرورية

```python
hourly = df.groupby("Hour")["Total"].mean().reset_index()
hourly


labels = ["12 AM","1 AM","2 AM","3 AM","4 AM","5 AM",
          "6 AM","7 AM","8 AM","9 AM","10 AM","11 AM",
          "12 PM","1 PM","2 PM","3 PM","4 PM","5 PM",
          "6 PM","7 PM","8 PM","9 PM","10 PM","11 PM"]

plt.figure(figsize=(14,5))
plt.plot(hourly.Hour, hourly.Total, marker="o")

plt.xticks(range(24), labels, rotation=45)
plt.title("Average Traffic by Hour")
plt.xlabel("Time")
plt.ylabel("Average Vehicles")
plt.grid(True) # هذا يطلع مربعات
plt.show()
```



### ما أكثر يوم ازدحامًا؟

```python
day_of_busy =df.groupby("Day of the week")["Total"].mean().reset_index()
```

الحركة المرورية مستقرة نسبيًا خلال الأسبوع، مع ارتفاع طفيف يوم السبت وانخفاض يوم الأحد.

```python
plt.figure(figsize=(10,5))
order = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
plt.xticks(ticks=range(len(order)), labels=order)
plt.bar(day_of_busy["Day of the week"],day_of_busy["Total"])
plt.title("Average Traffic by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Average Total")
plt.ylim(105,111)
```

## ما أكثر أنواع المركبات استخدامًا وقت الزحام؟

```python
vehicle_totals = df[["CarCount","BikeCount","BusCount","TruckCount"]].sum().reset_index()
vehicle_totals.columns =["vehicle","Total"]

vehicle_totals
```

```python
plt.bar(vehicle_totals.vehicle,vehicle_totals.Total)
plt.title("Total Vehicles by Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Count")
plt.show()
```

أي نوع مركبة يكون أعلى في أيام الأسبوع؟

```python
vehicle_by_day = df.groupby("Day of the week")[["CarCount","BikeCount","BusCount","TruckCount"]].mean().reset_index()
vehicle_by_day
```

```python
vehicle_by_day.plot(kind="bar")
plt.xticks(
    ticks=range(7),
    labels=["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"],
    rotation=45
)
plt.title("Average Vehicle Count by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Average Count")
plt.show()
```

## ما أكثر حالات المرور تكرارًا؟

```python
traffic_status = df["Traffic Situation"].value_counts().reset_index()
traffic_status.columns = ["Situation","Count"]

traffic_status.plot(kind="bar", x="Situation", y="Count", legend=False)

plt.title("Traffic Situation Count")
plt.xlabel("Situation")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

```python
heavy_by_hour =df[df["Traffic Situation"]=="heavy"].groupby("Hour").size().reset_index(name="Heavy Count")

plt.figure(figsize=(12,5))
plt.bar(heavy_by_hour["Hour"], heavy_by_hour["Heavy Count"])

labels = ["12 AM","1 AM","2 AM","3 AM","4 AM","5 AM",
          "6 AM","7 AM","8 AM","9 AM","10 AM","11 AM",
          "12 PM","1 PM","2 PM","3 PM","4 PM","5 PM",
          "6 PM","7 PM","8 PM","9 PM","10 PM","11 PM"]

plt.xticks(range(24), labels, rotation=45)

plt.title("Heavy Traffic by Hour")
plt.xlabel("Time")
plt.ylabel("Heavy Count")
plt.show()
```

## ما أنواع المركبات الأكثر مساهمة في الازدحام المروري الشديد؟

```python
heavy_reason = df[df["Traffic Situation"]=="heavy"][["CarCount","BikeCount","BusCount","TruckCount"]].mean().reset_index()
heavy_reason.columns = ["Vehicle Type", "Average Count"]
heavy_reason
```

```python
plt.figure(figsize=(8,5))

heavy_reason.plot(
    kind="bar",
    x="Vehicle Type",
    y="Average Count",
    legend=False,
    color="steelblue"
)

plt.title("Average Vehicles During Heavy Traffic")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Count")
plt.xticks(rotation=0)
plt.show()
```

## ما العلاقة بين عدد المركبات وحالة المرور؟

```python
df.groupby("Traffic Situation")["Total"].mean().reset_index()
```

```python
traffic_avg = df.groupby("Traffic Situation")["Total"].mean().reset_index()

plt.bar(traffic_avg["Traffic Situation"], traffic_avg["Total"])
plt.title("Average Vehicles by Traffic Situation")
plt.xlabel("Traffic Situation")
plt.ylabel("Average Vehicles")
plt.show()
```

## مقارنة أيام العمل ونهاية الأسبوع


```python
work_days = ["Sunday","Monday","Tuesday","Wednesday","Thursday"]
weekend = ["Friday","Saturday"]

df["DayType"] = df["Day of the week"].apply(
    lambda x: "Work Days" if x in work_days else "Weekend"
)
day_compare = df.groupby("DayType")["Total"].mean().reset_index()

day_compare
```

```python
plt.bar(day_compare["DayType"], day_compare["Total"])
plt.title("Traffic: Work Days vs Weekend")
plt.ylabel("Average Vehicles")
plt.show()
```
